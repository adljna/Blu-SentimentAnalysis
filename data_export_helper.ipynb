# 🔗 Data Export Helper

Jalankan notebook ini **setelah** Feature Engineering selesai.

Script ini membantu menyambungkan output dari tahap sebelumnya (Scraping → Preprocessing → Feature Engineering) ke folder `data/` yang digunakan oleh Modeling, Evaluation, dan Visualization.

**Pastikan kamu sudah punya file:**
- Output CSV dari notebook Feature Engineering

```python
import pandas as pd
import numpy as np
import os

os.makedirs('../data', exist_ok=True)
print('📁 Folder data/ siap.')
```

```python
# ============================================================
#  SESUAIKAN PATH INI dengan output Feature Engineering kamu!
# ============================================================
INPUT_PATH = '../Feature Engineering/blu_feature_engineered.csv'  # ← ganti jika beda
OUTPUT_PATH = '../data/blu_feature_engineered.csv'

df = pd.read_csv(INPUT_PATH)
print(f'Data dimuat: {df.shape}')
print(f'Kolom: {df.columns.tolist()}')
df.head()
```

```python
# ===========================================================
# Buat kolom 'sentiment' dari rating jika belum ada
# Asumsi kolom rating = 'score'
# ===========================================================
RATING_COL = 'score'  # ← sesuaikan jika beda

if 'sentiment' not in df.columns:
    if RATING_COL in df.columns:
        def label_from_rating(r):
            if r >= 4: return 'positif'
            elif r == 3: return 'netral'
            else: return 'negatif'
        df['sentiment'] = df[RATING_COL].apply(label_from_rating)
        print('✅ Kolom sentiment dibuat dari rating.')
    else:
        print(f'⚠️ Kolom "{RATING_COL}" tidak ada. Buat kolom sentiment secara manual.')
else:
    print('✅ Kolom sentiment sudah ada.')

print()
print('Distribusi sentimen:')
print(df['sentiment'].value_counts())
```

```python
# ===========================================================
# Pastikan ada kolom cleaned_text
# ===========================================================
POSSIBLE_TEXT_COLS = ['cleaned_text', 'text_clean', 'preprocessed', 'content_clean', 'content']
found_text_col = None

for col in POSSIBLE_TEXT_COLS:
    if col in df.columns:
        found_text_col = col
        break

if found_text_col and found_text_col != 'cleaned_text':
    df['cleaned_text'] = df[found_text_col]
    print(f'✅ Kolom "{found_text_col}" di-rename ke "cleaned_text".')
elif found_text_col == 'cleaned_text':
    print('✅ Kolom cleaned_text sudah ada.')
else:
    print('⚠️ Kolom teks bersih tidak ditemukan. Cek nama kolom di atas!')
    print(f'   Kolom tersedia: {df.columns.tolist()}')
```

```python
# Simpan ke folder data/
df.to_csv(OUTPUT_PATH, index=False)
print(f'✅ Data disimpan ke: {OUTPUT_PATH}')
print(f'   Shape: {df.shape}')
print()
print('Sekarang kamu bisa jalankan:')
print('  → Modeling/modeling.ipynb')
print('  → (setelah modeling) Evaluation/evaluation.ipynb')
print('  → (setelah evaluation) Visualization/visualization.ipynb')
```
